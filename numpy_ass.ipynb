```python
# 21. select()
# 61.	Given marks = [95, 82, 74, 61, 45, 30], use np.select() to assign A+ for >=90, A for >=80, B for >=70, C for >=60, and Fail otherwise.
import numpy as np
marks = np.array([95, 82, 74, 61, 45, 30])

conditions = [
    marks >= 90,
    marks >= 80,
    marks >= 70,
    marks >= 60
]

choices = ["A+", "A", "B", "C"]
result = np.select(conditions, choices, default="Fail")

print(result)

# 62.	Given sales = [500, 1200, 2500, 800, 3000, 1500], categorize them as High (>=2500), Medium (>=1000), and Low (<1000).
sales = np.array([500, 1200, 2500, 800, 3000, 1500])
conditions = [
    sales >= 2500,
    sales >= 1000
]

choices = ["High", "Medium"]
result = np.select(conditions, choices, default="Low")

print(result)

# 63.	Given age = [12, 17, 22, 35, 65, 70], create Minor (<18), Adult (18–59), and Senior (>=60) categories using np.select().
age = np.array([12, 17, 22, 35, 65, 70])

conditions = [
    age < 18,
    age < 60
]

choices = ["Minor", "Adult"]
result = np.select(conditions, choices, default="Senior")

print(result)
```

```python
# 35. Practical Student Analysis
# 103.	Given students = [[101,78,85,90],[102,65,72,80],[103,92,88,95],[104,45,60,55],[105,85,90,88]], where columns are Student_ID, 
# Maths, Science, English, calculate the total marks for each student.
import numpy as np
students = np.array([
    [101, 78, 85, 90],
    [102, 65, 72, 80],
    [103, 92, 88, 95],
    [104, 45, 60, 55],
    [105, 85, 90, 88]
])
marks = students[:, 1:]

total = np.sum(marks, axis=1)

print(total)

# 104.	Find the average marks of each student and identify the Student_ID of the student with the highest average.
marks = students[:, 1:]

average = np.mean(marks, axis=1)

print(average)

highest_index = np.argmax(average)

print(average[highest_index])
print(students[highest_index, 0])

# 105.	Using conditions, assign grades A for >=80, B for >=60, C for >=40, and Fail below 40 based on each student's average
marks = students[:, 1:]

average = np.mean(marks, axis=1)

conditions = [
    average >= 80,
    average >= 60,
    average >= 40
]

choices = ["A", "B", "C"]

grades = np.select(conditions, choices, default="Fail")

print(grades)
```
