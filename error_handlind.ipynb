```python
#error handling
try :                #risky code
    a = 10           
    b = 0
    print (a / b)
except :             #error hone per ye chalega
    print ("something went wrong")
```

```python
#example
try :
    a = 10
    b = 2
    print (a / b)
except :
    print ("something went wrong")
```

```python
#specific error handle karna 
try :
    a = 10
    b = 0
    print (a / b)
except ZeroDivisionError:
    print ("cannot divide by zero")
```

```python
#value error
try :
    age = int (input("enter age:"))
    print (age)
except ValueError :
    print ("please enter a number")
```
