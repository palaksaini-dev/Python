```python
# select  # np.select (conditions, choices, default="Fail")

import numpy as np 

marks = np.array ([85, 72, 45, 30, 90])

conditions = [
    marks >= 80,
    marks >= 60,
    marks >= 40
]

choices = [
    "A",
    "B",
    "C"
]

result = np.select (conditions, choices, default="Fail")

print (result)
```
