```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Aman", "Priya", "Rahul", "Neha", "Aman", "Priya", "Rahul", "Neha"],
    "City" : ["Jaipur", "Sikar", "Sikar", "Delhi", "Jaipur", "Sikar", "Sikar", "Jaipur"],
    "Department": ["IT", "HR", "IT", "HR", "IT", "HR", "Sales", "Sales"],
    "Salary": [25000, 30000, 35000, 28000, 40000, 32000, 27000, 36000],
    "Age": [22, 25, 24, 23, 26, 27, 22, 24]
})

print(df)
```

```python
df.groupby(['Department', 'City'])['Salary'].sum()
```

```python
a = df.groupby(['Department', 'City'])['Salary'].agg(['sum', 'mean', 'max'])
a
```

```python
df.groupby(['Department', 'City']).agg({
    "Salary": "sum",
    "Salary": "mean"
})
```

```python
import pandas as pd

df1 = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Palak", "Priya", "Akshya", "Ayush", "Neha"],
    "Branch": ["CSE", "IT", "CSE", "AI", "IT"]
})

print(df1)
```
