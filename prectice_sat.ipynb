```python
for i in range(5):

    # P
    for j in range(4):
        if j == 0 or (i == 0 and j < 3) or (i == 2 and j < 3) or (j == 3 and i == 1):
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print("   ", end=" ")

    # A
    for j in range(5):
        if (i == 0 and j == 2) or (i == 1 and (j == 1 or j == 3)) or i == 2 or (j == 0 or j == 4):
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print("   ", end=" ")

    # L
    for j in range(4):
        if j == 0 or i == 4:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print("   ", end=" ")

    # A
    for j in range(5):
        if (i == 0 and j == 2) or (i == 1 and (j == 1 or j == 3)) or i == 2 or j == 0 or j == 4:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print("   ", end=" ")

    # K
    for j in range(5):
        if j == 0:
            print("*", end=" ")
        elif i == 0 and j == 4:
            print("*", end=" ")
        elif i == 1 and j == 3:
            print("*", end=" ")
        elif i == 2 and j == 2:
            print("*", end=" ")
        elif i == 3 and j == 3:
            print("*", end=" ")
        elif i == 4 and j == 4:
            print("*", end=" ")
        else:
            print(" ", end=" ")

    print()
```

```python
num = [2, 7, 5, 4, 10, 6]

largest = num [0]
second = num [0]

for i in num :
    if i > largest:
        second = largest
        largest = i
    elif i > second and i != largest :
        second = i 
        print ("second largest:", second)
```

```python
n = 4321
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

print(rev)
```
