```python
#questions
product = {
    "apple" : 100, 
    "banana": 40,
    "milk" : 60, 
    "bread": 50,
    "eggs": 70
}

cart = [
    ("apple", 2),
    ("milk", 1),
    ("bread", 3)
]

subtotal = 0 
print ("GROCERY BILLING")
for item, quantity in cart :
    if item in product :
        price = product [item]
        cost = price * quantity
        subtotal = subtotal + cost 
        print("Item:", item)
        print("Quantity:", quantity)
        print("Cost: ", cost)
        print()

    else:
        print("item is not available")

tax =  subtotal * 0.05

if subtotal > 1000:
    discount = subtotal * 0.10
else:
    discount = 0

final_amount = subtotal + tax - discount 

print ("subtotal:", subtotal)
print ("tax (%5)", tax)
print ("discount ", discount )
print ("final_amount", final_amount)
```

```python
student = {
    "Alice": [85, 92, 78],
    "Bob": [70, 60, 65],
    "Charlie": [90, 95, 88],
    "David": [55, 60, 52]
}

grade_count = {
    "A": 0,
    "B": 0,
    "C": 0,
    "D": 0,
    "F": 0
}

for name, marks in student.items ():
    average = sum(marks) / len(marks)

    if average >= 90:
        grade = "A"
    elif average >= 80:
        grade = "B"
    elif average >= 70:
        grade = "C"
    elif average >= 60:
        grade = "D"
    else:
        grade = "F"

    grade_count [grade] += 1

    print("Name:", name)
    print("Average:", average)
    print("Grade:", grade)
    print()

for grade, count in grade_count.items():
    print("Grade", grade, ":", count, "student(s)")

```

```python
#lambda function -- lambda function ek short function hota hai, normal function ham def se banate hai, lekin chhote kaam ke liye lambda use kar sakte hai
def sq (x):
    return x * x
print (sq(5))
```

```python
sq = lambda x : x ** 2
print (sq(5))
```

```python
#Filter () function -- list me se sirf vo values choose karta hai jo condition satisfy karti hai 

num = [1, 2, 3, 4, 5, 6]
even_num = filter (lambda x : x  % 2 == 0 , num)
odd_num = filter (lambda x : x % 2 != 0, num)
print (list (even_num))
print (list (odd_num))
```

```python
#Create a lambda function to find the greater of two numbers.
greater = lambda a, b: a if a > b else b

print(greater(10, 20))

```

```python
#Create a lambda function that returns "Pass" if marks are 40 or above, otherwise "Fail".
result = lambda marks: "Pass" if marks >= 40 else "Fail"

print(result(55))
print(result(35))
```

```python
#Using map(), convert ["python", "sql", "power bi"] into uppercase.

subject = ["python", "sql", "power bi"]
uppercase_subject = list(map(lambda subject: subject.upper(), subject))
print (list(uppercase_subject))
```

```python
#Using map(), add 5 marks to every value: marks = [35, 50, 65, 80].
marks = [35, 50, 65, 80]

updated_marks = list(map(lambda mark: mark + 5, marks))

print(updated_marks)
```

```python
#Using filter(), select names starting with "A": names = ["Aman", "Riya", "Ankit", "Neha", "Anu"].


name = ["Aman", "Riya", "Ankit", "Neha", "Anu"]
result = list(filter(lambda name: name.startswith("A") , name))

print(result)
```

```python
#Using filter(), select prices between 500 and 1000.
price = [100, 200, 500, 700, 1000]

selected_price = list(filter(lambda price: 500 <= price <= 1000, price))

print(selected_price)
```
