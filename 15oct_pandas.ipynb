```python
#  Find the percentage of missing values in each column.
df.isnull().mean() * 100
```

```python
# Fill missing Age values with the average age.
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
# Fill missing Embarked values with the most common value.
df["Embarked"] = df["Embarked"].fillna(
    df["Embarked"].mode()[0]
)
```

```python
# Remove the Cabin column.
df = df.drop(columns=["Cabin"])
# df_clean = df.dropna()
```

```python
# Remove all rows containing missing values.
df_clean = df.dropna()
```

```python
# Find the survival percentage of male and female passengers.
df.groupby("Sex")["Survived"].mean() * 100
```

```python
# Find the top 5 passengers who paid the highest fares.
df.nlargest(5, "Fare")[["Name", "Fare"]]
```

```python
# Find the average fare of passengers who survived.
df[df["Survived"] == 1]["Fare"].mean()
```

```python
# Find the average age of passengers who did not survive.
df[df["Survived"] == 0]["Age"].mean()
```

```python
# Create a summary showing passenger count, average age, and survival rate for each passenger class.
df.groupby("Pclass").agg(
    passenger_count=("PassengerId", "count"),
    average_age=("Age", "mean"),
    survival_rate=("Survived", "mean")
)
```
