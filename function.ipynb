```python
#operations of art 
def calculator(a, b, operation):
    if operation == "+":
        return a + b
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "/":
        return a / b
    else:
        return "Invalid operation"


print(calculator(10, 5, "+"))
print(calculator(10, 5, "-"))
print(calculator(10, 5, "*"))
print(calculator(10, 5, "/"))
print(calculator(10, 5, "%"))
```

```python
#tem. que.
def temperature(temp):
    if temp > 35:
        return "Hot"
    elif temp >= 20:
        return "Normal"
    else:
        return "Cold"


print(temperature(40))
print(temperature(25))
print(temperature(15))
```

```python
#letter que.
def check_character(ch):
    if ch.isdigit():
        return "Digit"
    elif ch.isalpha():
        if ch.lower() in "aeiou":
            return "Vowel"
        else:
            return "Consonant"
    else:
        return "Special character"


print(check_character("a"))
print(check_character("b"))
print(check_character("5"))
print(check_character("@"))
```

```python
#triangle que.
def triangle(a, b, c):
    if a == b and b == c:
        return "Equilateral"
    elif a == b or b == c or a == c:
        return "Isosceles"
    else:
        return "Scalene"


print(triangle(5, 5, 5))
print(triangle(5, 5, 3))
print(triangle(5, 4, 3))
```
